```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# Create dataset
data = {
    "text": [
        "The teacher explained the topic very clearly",
        "I really enjoyed the class",
        "The lessons were interesting and easy to understand",
        "The teacher was very helpful",
        "The course content was excellent",
        "The class was boring and difficult to follow",
        "I did not understand the explanation",
        "The lessons were confusing",
        "The teacher did not explain the topic properly",
        "The course was difficult and boring"
    ],
    "label": [
        "positive",
        "positive",
        "positive",
        "positive",
        "positive",
        "negative",
        "negative",
        "negative",
        "negative",
        "negative"
    ]
}

# Convert data into DataFrame
df = pd.DataFrame(data)

print("Dataset:")
print(df)

# Separate text and labels
X = df["text"]
y = df["label"]

# Convert text into numerical features using TF-IDF
vectorizer = TfidfVectorizer()
X_tfidf = vectorizer.fit_transform(X)

# Split data into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf,
    y,
    test_size=0.2,
    random_state=42
)

# Create SVM model
model = SVC(kernel="linear")

# Train the model
model.fit(X_train, y_train)

# Predict test data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nActual Values:")
print(y_test.values)

print("\nPredicted Values:")
print(y_pred)

print("\nAccuracy:", accuracy)

# Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Test new sentences
new_text = [
    "The teacher explained everything clearly",
    "The class was very confusing"
]

new_text_tfidf = vectorizer.transform(new_text)
prediction = model.predict(new_text_tfidf)

print("\nNew Text Predictions:")

for text, result in zip(new_text, prediction):
    print(text, "->", result)
```
