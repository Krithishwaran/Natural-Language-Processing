```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# ---------------------------------------
# NLP Task 11: Text Classification using SVM
# ---------------------------------------

# Create dataset
data = {
    "text": [
        "The football team won the championship",
        "The player scored two goals in the final match",
        "India won the cricket match yesterday",
        "The coach announced the team for the tournament",
        "The batsman scored a century in the match",
        "The tennis player reached the final",
        "The basketball team won the league game",
        "The athlete won a gold medal",
        "The cricket team started training for the tournament",
        "The goalkeeper made an excellent save",

        "The new smartphone has a powerful processor",
        "Artificial intelligence is changing modern technology",
        "The company launched a new laptop",
        "Cloud computing helps companies store data",
        "The new software update improves security",
        "Machine learning is used in many applications",
        "The technology company developed a new processor",
        "Cybersecurity protects computer systems from attacks",
        "The new smartphone comes with an advanced camera",
        "Artificial intelligence can analyze large amounts of data"
    ],

    "label": [
        "sports",
        "sports",
        "sports",
        "sports",
        "sports",
        "sports",
        "sports",
        "sports",
        "sports",
        "sports",

        "technology",
        "technology",
        "technology",
        "technology",
        "technology",
        "technology",
        "technology",
        "technology",
        "technology",
        "technology"
    ]
}

# Convert into DataFrame
df = pd.DataFrame(data)

print("Dataset:")
print(df)

# ---------------------------------------
# Separate input and output
# ---------------------------------------

X = df["text"]
y = df["label"]

# ---------------------------------------
# Convert text into numerical features
# using TF-IDF
# ---------------------------------------

vectorizer = TfidfVectorizer()

X_tfidf = vectorizer.fit_transform(X)

print("\nTF-IDF Shape:", X_tfidf.shape)

# ---------------------------------------
# Split data into training and testing
# ---------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

# ---------------------------------------
# Create SVM model
# ---------------------------------------

model = SVC(kernel="linear")

# Train the model
model.fit(X_train, y_train)

print("\nSVM model trained successfully!")

# ---------------------------------------
# Make predictions
# ---------------------------------------

y_pred = model.predict(X_test)

print("\nActual Labels:")
print(y_test.values)

print("\nPredicted Labels:")
print(y_pred)

# ---------------------------------------
# Calculate accuracy
# ---------------------------------------

accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)

# ---------------------------------------
# Classification Report
# ---------------------------------------

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# ---------------------------------------
# Test new sentences
# ---------------------------------------

new_text = [
    "The football player scored a goal",
    "The company launched a new computer",
    "The cricket team won the match",
    "Machine learning is used in software"
]

# Convert new text into TF-IDF
new_text_tfidf = vectorizer.transform(new_text)

# Predict
predictions = model.predict(new_text_tfidf)

print("\nNew Text Predictions:")

for text, prediction in zip(new_text, predictions):
    print(text, "->", prediction)
```
