```python
# NLP Task 9: Laplace (Add-1) Smoothing

from collections import Counter

# Sample text
text = "I love NLP and I love Python"

# Convert text to lowercase and tokenize
words = text.lower().split()

# Count word frequencies
word_counts = Counter(words)

# Total number of words
total_words = len(words)

# Vocabulary size
vocabulary_size = len(word_counts)

print("Word Counts:")
print(word_counts)

# Laplace (Add-1) Smoothing
print("\nLaplace Smoothed Probabilities:")

for word in word_counts:
    probability = (word_counts[word] + 1) / (total_words + vocabulary_size)
    print(word, ":", round(probability, 4))

# Probability for an unseen word
unseen_word_probability = 1 / (total_words + vocabulary_size)

print("\nProbability of unseen word 'data':",
      round(unseen_word_probability, 4))
```
